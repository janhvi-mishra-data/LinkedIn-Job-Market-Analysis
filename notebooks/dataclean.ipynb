```python
skills_list = [
    'python',
    'sql',
    'power bi',
    'tableau',
    'excel',
    'java',
    'javascript',
    'html',
    'css',
    'react',
    'node.js',
    'aws',
    'azure',
    'machine learning',
    'deep learning',
    'tensorflow',
    'pandas',
    'numpy',
    'git',
    'postgresql'
]
```

```python
for skill in skills_list:
    df[skill] = df['description'].str.lower().str.contains(skill, na=False)
```

```python
skill_counts = {}

for skill in skills_list:
    skill_counts[skill] = df[skill].sum()

skill_df = pd.DataFrame(
    skill_counts.items(),
    columns=['Skill', 'Count']
)

skill_df.sort_values(
    by='Count',
    ascending=False,
    inplace=True
)

print(skill_df)
```

```python
import matplotlib.pyplot as plt

top_skills = skill_df.head(15)

plt.figure(figsize=(10,6))
plt.bar(top_skills['Skill'], top_skills['Count'])
plt.xticks(rotation=45)
plt.title('Top Skills in Indian Tech Jobs')
plt.show()
```

```python
city_mapping = {
    'Greater Bengaluru Area': 'Bengaluru',
    'Bengaluru North': 'Bengaluru',
    'Bengaluru East': 'Bengaluru',
    'Bangalore Urban': 'Bengaluru',
    'Gurgaon': 'Gurugram',
    'Mumbai Metropolitan Region': 'Mumbai',
    'Navi Mumbai': 'Mumbai',
    'Greater Delhi Area': 'Delhi',
    'New Delhi': 'Delhi',
    'Delhi Cantonment' : 'Delhi',
    'Pune/Pimpri-Chinchwad Area': 'Pune',
    'Greater Kolkata Area': 'Kolkata',
    'Thane': 'Mumbai',
    'Banjara Hills': 'Hyderabad',
    'Trivandrum': 'Thiruvananthapuram',
    'Jalandhar I': 'Jalandhar'
}

df['city'] = df['city'].replace(city_mapping)
```

```python
invalid_locations = [
    'India',
    'Karnataka',
    'Himachal Pradesh',
    'Rajasthan',
    'Andhra Pradesh'
]

df['city'] = df['city'].replace(invalid_locations, 'Unknown')
```

```python
company_summary = (
    df.groupby('companyName')
      .agg(
          Total_Jobs=('title','count'),
          Avg_Applications=('applicationsCount','mean')
      )
      .reset_index()
)
```

```python
city_summary = (
    df.groupby('city')
      .agg(
          Total_Jobs=('title','count')
      )
      .reset_index()
      .sort_values('Total_Jobs', ascending=False)
)
```

```python
metro_mapping = {
    'Bengaluru': 'Bengaluru Metro',
    'Mumbai': 'Mumbai Metro',
    'Pune': 'Pune Metro',
    'Delhi': 'Delhi NCR',
    'Noida': 'Delhi NCR',
    'Gurugram': 'Delhi NCR',
    'Hyderabad': 'Hyderabad Metro',
    'Chennai': 'Chennai Metro',
    'Kolkata': 'Kolkata Metro'
}

df['Metro_Region'] = df['city'].map(metro_mapping)
df['Metro_Region'] = df['Metro_Region'].fillna('Other')
```

```python
experience_summary = (
    df.groupby('experienceLevel')
      .agg(
          Total_Jobs=('title','count'),
          Avg_Applications=('applicationsCount','mean')
      )
      .reset_index()
      .sort_values('Total_Jobs', ascending=False)
)

experience_summary
```

```python
sector_summary = (
    df.groupby('sector')
      .agg(
          Total_Jobs=('title','count')
      )
      .reset_index()
      .sort_values('Total_Jobs', ascending=False)
)

sector_summary.head(15)
```

```python
metro_summary = (
    df.groupby('Metro_Region')
      .agg(
          Total_Jobs=('title','count')
      )
      .reset_index()
      .sort_values('Total_Jobs', ascending=False)
)

metro_summary
```

```python
title_summary = (
    df.groupby('title')
      .agg(
          Total_Jobs=('title','count')
      )
      .reset_index()
      .sort_values('Total_Jobs', ascending=False)
)

title_summary.head(20)
```

```python
monthly_summary = (
    df.groupby('Month')
      .agg(
          Total_Jobs=('title','count')
      )
      .reset_index()
)

monthly_summary
```

```python
import sqlite3

conn = sqlite3.connect("job_market.db")

df.to_sql(
    "jobs",
    conn,
    if_exists="replace",
    index=False
)
query = """
SELECT city,
       COUNT(*) AS Total_Jobs
FROM jobs
GROUP BY city
ORDER BY Total_Jobs DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT companyName,
       COUNT(*) AS Total_Jobs
FROM jobs
GROUP BY companyName
ORDER BY Total_Jobs DESC
LIMIT 10
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT experienceLevel,
       COUNT(*) AS Total_Jobs
FROM jobs
GROUP BY experienceLevel
ORDER BY Total_Jobs DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT experienceLevel,
       ROUND(AVG(applicationsCount), 2) AS AVG_Applications
FROM jobs
GROUP BY experienceLevel
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Metro_Region,
       COUNT(*) AS Total_Jobs
FROM jobs
GROUP BY Metro_Region
ORDER BY Total_Jobs DESC
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT city,
       ROUND(AVG(applicationsCount), 2) AS AVG_Applications
FROM jobs
GROUP BY city
ORDER BY AVG_Applications DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT sector,
       COUNT(*) AS Total_Jobs
FROM jobs
GROUP BY sector
ORDER BY Total_Jobs DESC;
"""

pd.read_sql_query(query, conn)
```
